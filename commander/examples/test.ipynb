```python
# import zeromq
import zmq
import json
```

```python
# create tcp socket
context = zmq.Context()
socket = context.socket(zmq.REQ)
socket.connect("tcp://pyxis:4199")
```

```python
# send message
parameters = [2,3]
msg = f'add {json.dumps(parameters)}'
msg = f'add 2,3'
```

```python
print(msg)
socket.send_string(msg)
# recv and json load
result = json.loads(socket.recv_string())
print(result)
```

```python
parameters = [{'a' : 1, 'b' : 2.5, 'c' : 'hello'}, 42]

msg = f'set_configuration {json.dumps(parameters)[1:-1]}'

socket.send_string(msg)

result = json.loads(socket.recv_string())
print(result)
```
